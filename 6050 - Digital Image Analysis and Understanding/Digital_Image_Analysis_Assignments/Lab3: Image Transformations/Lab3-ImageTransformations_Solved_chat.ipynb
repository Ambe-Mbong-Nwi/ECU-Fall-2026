#### Preliminary Code

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# Set the default figure size for all plots
plt.rcParams['figure.figsize'] = [10, 8]

def compose(frame, image, transformation):
    
    width, height = frame.size
    
    #Invert matrix for compose function, grab values for Affine Transform
    t = np.linalg.inv(transformation)
    a=t[0,0]; b=t[0,1]; c=t[0,2]; d=t[1,0]; e=t[1,1]; f=t[1,2]
    
    image = image.transform((width,height), Image.AFFINE,(a,b,c,d,e,f), Image.BICUBIC)

    #Make mask from image's location
    im = np.sum(np.asarray(image), -1)
    vals = 255.0*( im > 0)
    mask = Image.fromarray(vals).convert("1")

    #Composite images together
    result = Image.composite(image,frame,mask)

    return result
```


#### Solution: Building the 11 Missing Transformation Matrices

The screenshot gives three things for each picture:

1. **Scale** — how much bigger/smaller the picture should be.
2. **Rotation** — the angle in degrees.
3. **Translation** — the `(x, y)` position of the picture.

We use homogeneous coordinates, so a transformation has the form

\[
T = \begin{bmatrix}
a & b & t_x\\
c & d & t_y\\
0 & 0 & 1
\end{bmatrix}.
\]

The important detail in this lab is that image coordinates have **positive y going downward**. Therefore, for a positive angle, we use

\[
R(\theta)=
\begin{bmatrix}
\cos\theta & \sin\theta & 0\\
-\sin\theta & \cos\theta & 0\\
0&0&1
\end{bmatrix}.
\]

This makes a positive angle appear counter-clockwise in the displayed image, matching the angles in the supplied target collage.

We combine the operations as

\[
M = T_{\text{translation}}\,R_{\text{rotation}}\,S_{\text{scale}}.
\]

The order matters: scale first, then rotate, then translate.


```python

# ------------------------------------------------------------
# Helper function for creating a homogeneous transformation
# ------------------------------------------------------------

def make_transformation(scale=1.0, angle=0.0, tx=0.0, ty=0.0):
    """
    Create a 3x3 homogeneous transformation matrix.

    Parameters
    ----------
    scale : float
        Scaling factor.
    angle : float
        Rotation angle in degrees.
        Positive angles follow the image-coordinate convention
        used by this lab (positive y points downward).
    tx, ty : float
        Translation in x and y.

    Returns
    -------
    numpy.ndarray
        3x3 homogeneous transformation matrix.
    """

    # Convert degrees to radians because numpy.sin/cos use radians.
    theta = np.deg2rad(angle)

    # Scaling matrix.
    S = np.array([
        [scale, 0,     0],
        [0,     scale, 0],
        [0,     0,     1]
    ], dtype=float)

    # Rotation matrix for image coordinates.
    # The signs are opposite the usual mathematical rotation
    # matrix because image y increases downward.
    R = np.array([
        [np.cos(theta),  np.sin(theta), 0],
        [-np.sin(theta), np.cos(theta), 0],
        [0,              0,             1]
    ], dtype=float)

    # Translation matrix.
    T = np.array([
        [1, 0, tx],
        [0, 1, ty],
        [0, 0, 1]
    ], dtype=float)

    # Apply scale, then rotation, then translation.
    return T @ R @ S


# ------------------------------------------------------------
# Target values read from the supplied collage
#
# Bird0 was already provided by the instructor.
# The remaining Bird1-Bird11 values are the 11 target frames.
# ------------------------------------------------------------

transform_specs = {
    "Bird0.png":  {"scale": 1.000, "angle":   0, "tx": 619, "ty": 433},
    "Bird1.png":  {"scale": 1.210, "angle":   0, "tx":  41, "ty":  30},
    "Bird2.png":  {"scale": 0.389, "angle":   0, "tx": 283, "ty":  46},
    "Bird3.png":  {"scale": 1.000, "angle":  30, "tx": 419, "ty":  87},
    "Bird4.png":  {"scale": 0.653, "angle": -15, "tx": 673, "ty":  37},
    "Bird5.png":  {"scale": 1.000, "angle": -45, "tx": 350, "ty": 138},
    "Bird6.png":  {"scale": 0.389, "angle":   0, "tx": 514, "ty": 225},
    "Bird7.png":  {"scale": 0.736, "angle":  15, "tx": 633, "ty": 228},
    "Bird8.png":  {"scale": 1.210, "angle":  15, "tx":  46, "ty": 354},
    "Bird9.png":  {"scale": 0.389, "angle": -45, "tx": 305, "ty": 358},
    "Bird10.png": {"scale": 1.000, "angle":  45, "tx": 385, "ty": 379},
    "Bird11.png": {"scale": 0.736, "angle":   0, "tx": 308, "ty": 463},
}


# Create all 12 matrices.
transformations = {}

for filename, values in transform_specs.items():
    transformations[filename] = make_transformation(**values)

# Print the matrices so you can see the actual numerical results.
np.set_printoptions(precision=6, suppress=True)

for filename, matrix in transformations.items():
    print(f"{filename}:")
    print(matrix)
    print()
```

\[

The supplied `compose()` function then inverts this matrix before passing it to `PIL.Image.transform()`. That inversion is necessary because PIL's affine transform maps output coordinates back to the source image.


```python

# ------------------------------------------------------------
# Compose all 12 bird images onto the supplied frame.
# ------------------------------------------------------------

# Start with the blank picture-frame collage.
result = frame.copy()

# Add each bird using its transformation matrix.
for filename, transformation in transformations.items():

    # Open the bird image.
    image = Image.open(filename).convert("RGB")

    # Put the transformed image onto the collage.
    result = compose(result, image, transformation)


# Display the completed collage.
plt.figure(figsize=(10, 8))
plt.imshow(result)
plt.axis("off")
plt.show()

# Uncomment this if you want to save your finished answer.
# result.save("Output.png")
```
