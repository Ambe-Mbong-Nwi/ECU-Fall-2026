```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
im = plt.imread('cat.jpg')

plt.imshow(im, cmap="gray");plt.show()

print(im.shape)
```

```python
#pad_im = np.pad(im, ((1,1),(1,1)), 'edge')

rows, cols = im.shape

buffer_im = np.zeros((rows,cols))

### BAD
# for i in range(1,rows-1):
#     for j in range(1,cols-1):
#         for k in range(-1,2):
#             for l in range(-1,2):
#                 buffer_im[i,j] = im[i+k,j+l]

### GOOD
for i in range(1,rows-1):
    for j in range(1,cols-1):
        buffer_im[i,j] = np.mean(im[i-1:i+2,j-1:j+2])

# Lop off padding
#buffer_im[1:-1,1:-1]

print(buffer_im.shape)

plt.imshow(buffer_im, cmap="gray");plt.show()




```

```python
rows, cols = im.shape

output_im = np.zeros((rows,cols),dtype=np.float32)

for i in range(1,rows-1):
    for j in range(1,cols-1):
        output_im[i,j] += np.sum(1*im[i-1:i+2,j-1:j+2])

output_im /= 9

plt.imshow(output_im, cmap="gray");plt.show()
```
