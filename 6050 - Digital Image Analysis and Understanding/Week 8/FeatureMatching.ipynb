```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python
img1 = cv2.imread("left-stitch.jpg",cv2.IMREAD_GRAYSCALE)
img2 = cv2.imread("right-stitch.jpg",cv2.IMREAD_GRAYSCALE)

plt.imshow(img1,cmap="gray");plt.show()
plt.imshow(img2,cmap="gray");plt.show()
```

```python
sift = cv2.SIFT_create()

kp1, des1 = sift.detectAndCompute(img1,None)
kp2, des2 = sift.detectAndCompute(img2,None)
```

```python
bf = cv2.BFMatcher()
matches = bf.knnMatch(des1,des2,k=2)

good_matches = []
for m,n in matches:
    if m.distance < 0.75*n.distance:
        good_matches.append([m])

result = cv2.drawMatchesKnn(img1,kp1,img2,kp2,good_matches,None,flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS)
```

```python
plt.imshow(result);plt.show()
```
