```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
bird = cv2.imread("test.png")

print(bird.shape)
```

```python
#BGR vs RGB
plt.imshow(bird[:,:,::-1])
plt.show()
```

```python
# Don't use in Jupyter Notebook
cv2.imshow("Bird",bird)
cv2.waitKey(0)
cv2.deleteAllWindows()
```

```python
# Color Conversion
bird_rgb = cv2.cvtColor(bird,cv2.COLOR_BGR2RGB)
#plt.imshow(bird_rgb)
#plt.show()

bird_gray = cv2.cvtColor(bird,cv2.COLOR_BGR2GRAY)
#plt.imshow(bird_gray,cmap="gray")
#plt.show()

bird_hsv = cv2.cvtColor(bird,cv2.COLOR_BGR2HSV)
```

```python
# Convolution
kernel=np.array([[1,1,1],
                [1,1,1],
                [1,1,1]])/9

bird_blur = cv2.filter2D(bird,ddepth=-1,kernel=kernel)

bird_blur = cv2.blur(bird,ksize=(5,5))

plt.imshow(bird_blur[:,:,::-1])
plt.show()
```

```python
# Edge Detection

bird_edges = cv2.Canny(bird,20,240)

plt.imshow(bird_edges,cmap="gray")
plt.show()
```

```python
# Transformations

# Scaling
bird_big = cv2.resize(bird,(400,400),interpolation=cv2.INTER_CUBIC)

plt.imshow(bird_big[:,:,::-1])
plt.show()
```

```python
# Rotation

M = cv2.getRotationMatrix2D((72,72),-30,1)
bird_rot = cv2.warpAffine(bird,M,(144,144))

plt.imshow(bird_rot[:,:,::-1])
plt.show()
```

```python
# Perspective
pts1 = np.float32([[56,65],[100,52],[28,100],[120,140]])
pts2 = np.float32([[0,0],[144,0],[0,144],[144,144]])

M = cv2.getPerspectiveTransform(pts1,pts2)

bird_per = cv2.warpPerspective(bird,M,(144,144))

plt.imshow(bird_per[:,:,::-1])
plt.show()
```
